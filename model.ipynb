```python
# ===============================================================
# UNDERTONE CLASSIFIER — K-Fold Training + Ensemble + TTA
# ===============================================================

import os
import torch
import torch.nn as nn
import numpy as np
from torch.utils.data import DataLoader, Subset, ConcatDataset
from torchvision import datasets, transforms, models
from torchvision.models import resnet50, ResNet50_Weights
from sklearn.model_selection import KFold
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("🧠 Using device:", device)

# ===============================
# Download Dataset
# ===============================
dataset_path = kagglehub.dataset_download("gottumukkalamadhuri/undertone-dataset")
print("Path to dataset files:", dataset_path)

# ===============================
# Paths & Hyperparameters
# ===============================
train_dir = os.path.join(dataset_path, "train")
valid_dir = os.path.join(dataset_path, "valid")
test_dir  = os.path.join(dataset_path, "test")

num_classes  = 4
batch_size   = 32
epochs_fc    = 25
epochs_l34   = 50
lr_fc        = 1e-3
lr_l34       = 1e-4
patience     = 5
k_folds      = 5
alpha_mixup  = 0.4
min_delta    = 0.001

# ===============================
# Transforms
# ===============================
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

tta_transforms = [
    transforms.Compose([transforms.Resize((224,224)), transforms.ToTensor(),
                        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])]),
    transforms.Compose([transforms.Resize((224,224)), transforms.RandomHorizontalFlip(p=1.0),
                        transforms.ToTensor(),
                        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])]),
    transforms.Compose([transforms.Resize((224,224)), transforms.RandomRotation(15),
                        transforms.ToTensor(),
                        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])])
]

# ===============================
# Datasets
# ===============================
train_dataset = datasets.ImageFolder(train_dir, transform=None)
valid_dataset = datasets.ImageFolder(valid_dir, transform=None)
test_dataset  = datasets.ImageFolder(test_dir,  transform=test_transform)

full_dataset = ConcatDataset([train_dataset, valid_dataset])
classes = train_dataset.classes

# ===============================
# Subset Wrapper
# ===============================
class SubsetWithTransform(torch.utils.data.Dataset):
    def __init__(self, subset, transform=None):
        self.subset    = subset
        self.transform = transform

    def __getitem__(self, idx):
        x, y = self.subset[idx]
        if self.transform:
            x = self.transform(x)
        return x, y

    def __len__(self):
        return len(self.subset)

# ===============================
# MixUp
# ===============================
def mixup_data(x, y, alpha=0.4):
    lam = np.random.beta(alpha, alpha) if alpha > 0 else 1
    idx = torch.randperm(x.size(0)).to(x.device)
    return lam * x + (1 - lam) * x[idx], y, y[idx], lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

# ===============================
# Eval Helper
# ===============================
def eval_metrics(model, loader, dataset_name, classes, device):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            _, pred = torch.max(model(imgs), 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(pred.cpu().numpy())
    acc = 100 * np.sum(np.array(y_true) == np.array(y_pred)) / len(y_true)
    print(f"\n✅ {dataset_name} Accuracy: {acc:.2f}%")
    print(classification_report(y_true, y_pred, target_names=classes, digits=4))
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap="Blues", xticklabels=classes, yticklabels=classes)
    plt.title(f"Confusion Matrix - {dataset_name}")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.show()

# ===============================
# K-Fold Training
# ===============================
kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
fold_val_accuracies = []

for fold, (train_idx, val_idx) in enumerate(kf.split(range(len(full_dataset)))):
    print(f"\n==================== Fold {fold+1}/{k_folds} ====================")

    train_loader = DataLoader(
        SubsetWithTransform(Subset(full_dataset, train_idx), transform=train_transform),
        batch_size=batch_size, shuffle=True, num_workers=2
    )
    val_loader = DataLoader(
        SubsetWithTransform(Subset(full_dataset, val_idx), transform=test_transform),
        batch_size=batch_size, shuffle=False, num_workers=2
    )

    # Model
    model = resnet50(weights=ResNet50_Weights.IMAGENET1K_V1)
    for param in model.parameters():
        param.requires_grad = False
    model.fc = nn.Sequential(
        nn.Linear(model.fc.in_features, 512),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(512, num_classes)
    )
    model = model.to(device)

    # Class weights
    targets      = np.array([full_dataset[i][1] for i in train_idx])
    class_counts = [np.sum(targets == i) for i in range(num_classes)]
    class_weights = torch.tensor([sum(class_counts)/c for c in class_counts], dtype=torch.float).to(device)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    save_path = f"/kaggle/working/fold{fold+1}_best.pth"

    # ---- Step 1: FC only ----
    optimizer = torch.optim.AdamW(model.fc.parameters(), lr=lr_fc, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', patience=2, factor=0.3)
    best_val_loss, trigger = np.inf, 0

    for epoch in range(epochs_fc):
        model.train()
        train_loss, correct, total = 0, 0, 0
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            imgs, y_a, y_b, lam = mixup_data(imgs, labels, alpha_mixup)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = mixup_criterion(criterion, outputs, y_a, y_b, lam)
            loss.backward(); optimizer.step()
            train_loss += loss.item() * imgs.size(0)
            correct += (torch.max(outputs, 1)[1] == labels).sum().item()
            total   += labels.size(0)
        train_loss /= total; train_acc = 100 * correct / total

        model.eval()
        val_loss, correct, total = 0, 0, 0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * imgs.size(0)
                correct  += (torch.max(outputs, 1)[1] == labels).sum().item()
                total    += labels.size(0)
        val_loss /= total; val_acc = 100 * correct / total
        scheduler.step(val_loss)
        print(f"[FC] Epoch {epoch+1}/{epochs_fc} | Train Acc: {train_acc:.2f}% | Val Acc: {val_acc:.2f}% | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")

        if val_loss < best_val_loss - min_delta:
            best_val_loss = val_loss; trigger = 0
            torch.save(model.state_dict(), save_path)
        else:
            trigger += 1
            if trigger >= patience:
                print("Early stopping FC training..."); break

    # ---- Step 2: Layer3 + Layer4 fine-tune ----
    for name, param in model.named_parameters():
        if any(k in name for k in ["layer3", "layer4", "fc"]):
            param.requires_grad = True

    optimizer = torch.optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()), lr=lr_l34, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10)
    best_val_loss, trigger = np.inf, 0

    for epoch in range(epochs_l34):
        model.train()
        train_loss, correct, total = 0, 0, 0
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward(); optimizer.step()
            train_loss += loss.item() * imgs.size(0)
            correct += (torch.max(outputs, 1)[1] == labels).sum().item()
            total   += labels.size(0)
        train_loss /= total; train_acc = 100 * correct / total

        model.eval()
        val_loss, correct, total = 0, 0, 0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * imgs.size(0)
                correct  += (torch.max(outputs, 1)[1] == labels).sum().item()
                total    += labels.size(0)
        val_loss /= total; val_acc = 100 * correct / total
        scheduler.step()
        print(f"[Layer34] Epoch {epoch+1}/{epochs_l34} | Train Acc: {train_acc:.2f}% | Val Acc: {val_acc:.2f}% | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")

        if val_loss < best_val_loss - min_delta:
            best_val_loss = val_loss; trigger = 0
            torch.save(model.state_dict(), save_path)
        else:
            trigger += 1
            if trigger >= patience:
                print("Early stopping Layer3+Layer4 training..."); break

    # ---- Per-fold metrics ----
    model.load_state_dict(torch.load(save_path))
    eval_metrics(model, val_loader,  "Validation", classes, device)
    eval_metrics(model, test_loader, "Test",       classes, device)

    # Track val accuracy for ensemble weighting
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for imgs, labels in val_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            correct += (torch.max(model(imgs), 1)[1] == labels).sum().item()
            total   += labels.size(0)
    fold_val_accuracies.append(correct / total)

print("\n📌 Fold Val Accuracies:", [f"{a:.4f}" for a in fold_val_accuracies])

# ===============================
# Ensemble + TTA
# ===============================
model_paths = [f"/kaggle/working/fold{i+1}_best.pth" for i in range(k_folds)]
models_list = []

for path in model_paths:
    m = resnet50(weights=None)
    m.fc = nn.Sequential(
        nn.Linear(m.fc.in_features, 512),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(512, num_classes)
    )
    m.load_state_dict(torch.load(path, map_location=device))
    m = m.to(device)
    m.eval()
    models_list.append(m)

def ensemble_predict_tta(img_pil):
    weighted_probs = torch.zeros(num_classes).to(device)
    total_weight = len(tta_transforms) * sum(fold_val_accuracies)

    for tform in tta_transforms:
        img_tensor = tform(img_pil).unsqueeze(0).to(device)
        for m, w in zip(models_list, fold_val_accuracies):
            with torch.no_grad():
                probs = torch.softmax(m(img_tensor), dim=1).squeeze(0)
                weighted_probs += w * probs

    weighted_probs /= total_weight
    return torch.argmax(weighted_probs).item(), weighted_probs.cpu().numpy()

# ---- Evaluate ensemble on test set ----
y_true, y_pred = [], []
for cls_idx, cls_name in enumerate(classes):
    cls_folder = os.path.join(test_dir, cls_name)
    for img_name in os.listdir(cls_folder):
        img = Image.open(os.path.join(cls_folder, img_name)).convert("RGB")
        pred, _ = ensemble_predict_tta(img)
        y_true.append(cls_idx)
        y_pred.append(pred)

acc = accuracy_score(y_true, y_pred)
print(f"\n🏆 Ensemble Test Accuracy with TTA: {acc*100:.2f}%")
print("\n📊 Classification Report:\n", classification_report(y_true, y_pred, target_names=classes, digits=4))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap="Blues", xticklabels=classes, yticklabels=classes)
plt.title("Confusion Matrix - Ensemble + TTA")
plt.xlabel("Predicted"); plt.ylabel("True")
plt.show()

# ===============================
# Single Image Prediction
# ===============================
# Auto-picks the first image from the test set.
# Replace img_path with your own image path if needed.
sample_class_folder = os.path.join(test_dir, classes[0])
sample_img_name     = os.listdir(sample_class_folder)[0]
img_path            = os.path.join(sample_class_folder, sample_img_name)

img = Image.open(img_path).convert("RGB")
pred_class, probs = ensemble_predict_tta(img)

plt.figure(figsize=(5, 5))
plt.imshow(img)
plt.axis('off')
plt.title(f"Predicted: {classes[pred_class]}\nProbabilities: {np.round(probs, 3)}")
plt.show()
```
